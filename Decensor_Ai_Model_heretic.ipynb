```python
# @title 🚀 1. Decensor Model (All-in-One Smart Engine & Auto Resume) ⚡
# @markdown Enter any Hugging Face model! Checks GPU, auto-installs packages, mounts Drive, and resumes progress seamlessly.

MODEL_NAME = "Qwen/Qwen3-1.7B"  # @param {type:"string"}
QUANTIZATION = "BNB_4BIT"  # @param ["BNB_4BIT", "NONE"] {allow-input: true}
HUGGINGFACE_TOKEN = ""  # @param {type:"string"}

import os
import shutil
import torch

# -------------------------------------------------------------------
# 🔍 1. GPU Diagnostics & VRAM Sanity Check
# -------------------------------------------------------------------
print("🩺 Running Pre-Flight System Diagnostics...")

if not torch.cuda.is_available():
  print(
      "\n❌ CRITICAL ERROR: No GPU detected!"
      "\n👉 Please go to top menu: Runtime -> Change runtime type -> Select T4 GPU"
      " (or A100)."
  )
  raise SystemExit("🚨 Process paused: GPU hardware is required to run Heretic.")
else:
  gpu_name = torch.cuda.get_device_name(0)
  vram_gb = torch.cuda.get_device_properties(0).total_memory / (1024**3)
  print(
      f"✅ GPU Online: {gpu_name} ({vram_gb:.2f} GB VRAM available) - Ready to"
      " melt weights! 🔥\n"
  )

# -------------------------------------------------------------------
# 📁 2. Mount Google Drive & Isolate Model Checkpoints
# -------------------------------------------------------------------
print("📁 Connecting Google Drive...")
from google.colab import drive

drive.mount("/content/drive")

# Generate a unique checkpoint folder based on the model name
clean_model_name = MODEL_NAME.replace("/", "_")
CHECKPOINT_DIR = (
    f"/content/drive/MyDrive/heretic_checkpoints/{clean_model_name}"
)
os.makedirs(CHECKPOINT_DIR, exist_ok=True)
print("✅ Drive connected! Checkpoint folder ready.\n")

# -------------------------------------------------------------------
# ⚙️ 3. Auto-Install Heretic Engine & others (if missing from session)
# -------------------------------------------------------------------

print("🛠️ Heretic engine & torchao installing now...")
!pip install -U torchao
!pip install -q -U heretic-llm bitsandbytes accelerate huggingface_hub
print("🎉 Heretic & CUDA packages & torchao installed successfully!\n")



# -------------------------------------------------------------------
# 🔑 4. Hugging Face Authentication
# -------------------------------------------------------------------
clean_hf_token = HUGGINGFACE_TOKEN.strip()
if clean_hf_token:
    os.environ["HF_TOKEN"] = clean_hf_token
    !huggingface-cli login --token "$clean_hf_token"

# -------------------------------------------------------------------
# 🔥 5. Launch Heretic Engine!
# -------------------------------------------------------------------
print(f"\n🚀 Igniting Heretic on model: {MODEL_NAME}")
print(f"💾 Isolated Checkpoint Folder: {CHECKPOINT_DIR}\n")

!heretic --model "{MODEL_NAME}" --quantization {QUANTIZATION} --study-checkpoint-dir "{CHECKPOINT_DIR}"

print("\n✅ Decensoring process complete! You can move to the next cell to zip and download.")
```

```python
# @title 📦 2. Create Zip Archive
# @markdown Smart-scans Colab to find your new model folder and zips it up without missing a beat! 🎯

import os
import shutil

zip_output_path = "/content/uncensoredai"

# Smart search for the directory containing the new model files
model_dirs = []
for root, dirs, files in os.walk("/content"):
  if "config.json" in files and root != "/content":
    if ".cache" not in root and "huggingface" not in root:
      model_dirs.append(root)

if model_dirs:
  target_dir = model_dirs[0]
  print(f"🎯 Target model folder detected at: {target_dir}")
  print("⏳ Packing up model files into zip... (takes few minute)")
  shutil.make_archive(zip_output_path, "zip", target_dir)
  print(f"🎉 Zip created successfully: {zip_output_path}.zip")
else:
  print("⚠️ No subfolder found, zipping main workspace directory...")
  shutil.make_archive(zip_output_path, "zip", "/content")
  print("🎉 Zip created successfully!")
```
